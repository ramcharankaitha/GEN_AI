# RAG Step 2 — Storing Chunks in ChromaDB

This notebook stores the chunks produced in Step 1 inside a ChromaDB collection and demonstrates basic retrieval.

## 1. Install and import ChromaDB

```python
!pip -q install chromadb

import chromadb
print("ChromaDB imported successfully.")
```

## 2. Create the chunks

```python
chunks = [
    "RAG combines information retrieval with language generation.",
    "A RAG system stores documents in a searchable form.",
    "Relevant pieces of documents are retrieved when a user asks a question.",
    "Retrieved context is provided to a language model to generate a grounded answer.",
    "Chunking helps large documents become easier to retrieve efficiently."
]

for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i}: {chunk}")
```

## 3. Create a ChromaDB collection

```python
client = chromadb.Client()
collection = client.get_or_create_collection(name="rag_chunks")

print("Collection created:", collection.name)
```

## 4. Store the chunks

```python
ids = [f"chunk_{i}" for i in range(len(chunks))]

collection.add(
    documents=chunks,
    ids=ids
)

print("Stored documents:", collection.count())
```

## 5. Inspect stored chunks

```python
stored = collection.get()

for doc_id, doc in zip(stored["ids"], stored["documents"]):
    print(f"{doc_id}: {doc}")
```

## 6. Semantic query

```python
results = collection.query(
    query_texts=["How does RAG retrieve useful information?"],
    n_results=2
)

print("Query:", "How does RAG retrieve useful information?")
print("\nTop results:")
for i, doc in enumerate(results["documents"][0], 1):
    print(f"{i}. {doc}")
```

## 7. Final status

```python
print("ChromaDB storage: completed")
print("Collection:", collection.name)
print("Total stored chunks:", collection.count())
```
