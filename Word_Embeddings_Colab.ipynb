# Word Embeddings

This Colab notebook demonstrates **One-Hot Encoding** and **Word2Vec word embeddings** using `scikit-learn` and `gensim`.

```python
from sklearn.preprocessing import OneHotEncoder

words = [['cat'], ['dog'], ['apple'], ['car']]

encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)

print("words :", encoder.categories_[0])
print("Embeddings:")
print(result)
```

```python
from gensim.models import Word2Vec

sentences = [
    ["python", "is", "a", "language"],
    ["java", "is", "a", "language"],
    ["python", "is", "easy"],
    ["java", "is", "easy"],
    ["python", "is", "powerful"],
    ["java", "is", "powerful"]
]

model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, workers=1)

print(model.wv.most_similar('java', topn=3))
```

```python
print(model.wv.most_similar('python', topn=4))
```

## Conclusion

One-Hot Encoding represents each word as a sparse binary vector, while Word2Vec learns dense numerical representations from word context.
