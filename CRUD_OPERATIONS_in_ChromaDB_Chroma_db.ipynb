```python
%pip install -q chromadb

import chromadb

# STEP 1: CREATE CHROMADB CLIENT
client = chromadb.Client()

# STEP 2: DELETE EXISTING COLLECTION IF IT EXISTS
try:
    client.delete_collection(name="technology_demo")
except Exception:
    pass

# STEP 3: CREATE COLLECTION
collection = client.create_collection(name="technology_demo")

print("Collection created:", collection.name)
print("Initial document count:", collection.count())


# STEP 4: ADD DOCUMENTS
collection.add(
    documents=[
        "Machine learning enables computers to learn patterns from data.",
        "Python is widely used for artificial intelligence and data science.",
        "PostgreSQL is a relational database management system.",
        "Cloud computing provides computing resources over the internet.",
        "Cybersecurity protects computer systems, networks, and data."
    ],
    metadatas=[
        {"category": "AI"},
        {"category": "programming"},
        {"category": "database"},
        {"category": "cloud"},
        {"category": "security"}
    ],
    ids=["tech1", "tech2", "tech3", "tech4", "tech5"]
)

print("\nDocuments added!")
print("Total count:", collection.count())


# STEP 5: PREVIEW DOCUMENTS
print("\n--- PREVIEW DOCUMENTS ---")

preview = collection.peek(limit=3)

for doc_id, doc in zip(preview["ids"], preview["documents"]):
    print(doc_id, "->", doc)


# STEP 6: RETRIEVE DOCUMENTS BY ID
print("\n--- RETRIEVE DOCUMENTS ---")

result = collection.get(ids=["tech1", "tech3"])

for doc_id, doc, meta in zip(
    result["ids"],
    result["documents"],
    result["metadatas"]
):
    print(f"{doc_id} | {meta} | {doc}")


# STEP 7: SEMANTIC SEARCH
print("\n--- SEMANTIC SEARCH ---")

results = collection.query(
    query_texts=["How do computers learn from information?"],
    n_results=3
)

for doc_id, doc, distance in zip(
    results["ids"][0],
    results["documents"][0],
    results["distances"][0]
):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")


# STEP 8: SEARCH USING METADATA FILTER
print("\n--- DATABASE CATEGORY SEARCH ---")

results = collection.query(
    query_texts=["Where can I store structured information?"],
    n_results=3,
    where={"category": "database"}
)

for doc_id, doc, distance in zip(
    results["ids"][0],
    results["documents"][0],
    results["distances"][0]
):
    print(f"{doc_id} | distance={distance:.4f} | {doc}")


# STEP 9: UPDATE DOCUMENT
print("\n--- UPDATE DOCUMENT ---")

collection.update(
    ids=["tech4"],
    documents=[
        "Cloud computing provides scalable servers, storage, databases, and networking over the internet."
    ],
    metadatas=[
        {"category": "cloud", "updated": True}
    ]
)

updated = collection.get(ids=["tech4"])

print("Updated document:", updated["documents"])
print("Updated metadata:", updated["metadatas"])


# STEP 10: UPSERT DOCUMENTS
print("\n--- UPSERT DOCUMENTS ---")

collection.upsert(
    documents=[
        "Python supports machine learning, automation, and backend development.",
        "Docker packages applications and their dependencies into containers."
    ],
    metadatas=[
        {"category": "programming"},
        {"category": "devops"}
    ],
    ids=["tech2", "tech6"]
)

print("Total count after upsert:", collection.count())

result = collection.get(ids=["tech2", "tech6"])

for doc_id, doc in zip(result["ids"], result["documents"]):
    print(doc_id, "->", doc)


# STEP 11: DELETE DOCUMENT BY ID
print("\n--- DELETE DOCUMENT BY ID ---")

collection.delete(ids=["tech5"])

print("Count after deleting tech5:", collection.count())
print("Remaining IDs:", collection.get()["ids"])


# STEP 12: DELETE DOCUMENTS BY CATEGORY
print("\n--- DELETE BY CATEGORY ---")

collection.delete(
    where={"category": "programming"}
)

print("Count after category deletion:", collection.count())
print("Remaining IDs:", collection.get()["ids"])


# STEP 13: DISPLAY FINAL DATABASE STATE
print("\n--- FINAL DATABASE STATE ---")

final_state = collection.get()

print("Final document count:", collection.count())

for doc_id, doc, meta in zip(
    final_state["ids"],
    final_state["documents"],
    final_state["metadatas"]
):
    print(f"{doc_id} | {meta} | {doc}")


# STEP 14: FINAL VERIFICATION
print("\n--- VERIFICATION ---")

assert collection.count() == 4

print("All operations completed successfully!")
print("Expected final document count: 4")
print("Actual final document count:", collection.count())
```
